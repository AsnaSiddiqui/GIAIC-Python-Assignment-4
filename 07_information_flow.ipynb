```python
print("00_choosing_returns \n")

def choose_return(option):
    if option == "A":
        return "You choose option A"
    elif option == "B":
        return "You choose option B"
    elif option == "C":
        return "You choose option C"
    else:
        return "Invalid option. Please choose A, B, or C."

def main():
    print("Choose one of the following options: A, B, C")
    option = input("Enter your choice:  ")

    result = choose_return(option)
    print(result)

if __name__ == "__main__":
    main()
```

```python
print("02_in_range \n")

def is_in_range(number, start, end):
    if start <= number <= end:
        return True
    else:
        return False

def main():
    try:
        num = int(input("Enter a number: "))
        start = int(input("Enter the start of the range:  "))
        end = int(input("Enter the end of the range:  "))
    except ValueError:
        print("Please enter valid integers.")
        return

    if is_in_range(num, start, end):
        print(f"{num} is in the range of {start} and {end}.")
    else:
        print(f"{num} is not in the range of {start} and {end}.")

if __name__ == "__main__":
    main()
```

```python
print("03_in_stock \n")

def is_in_stock(product, stock):
    if product in stock and stock[product] > 0:
        return True
    else:
        return False

def main():
    stock = {
        "apple": 10,
        "banana": 0,
        "strawberry": 5,
        "mango": 2
    }

    product = input("Enter the product name to check its stock: ").lower()

    if is_in_stock(product, stock):
        print(f"{product.capitalize()} is in stock!")
    else:
        print(f"{product.capitalize()} is not in stock.")

if __name__ == "__main__":
    main()
```

```python
print("04_multiple_returns \n")

import math

def analyze_number(number):
    square = number ** 2
    cube = number ** 3
    square_root = math.sqrt(number)

    return square, cube, square_root

def main():
    try:
        num = float(input("Enter a number:  "))
    except ValueError:
        print("Please enter a valid number.")
        return

    square, cube, square_root = analyze_number(num)

    print(f"The square of {num} is: {square}")
    print(f"he cube of {num} is: {cube}")
    print(f"The square root of {num} is: {square_root:.2f}")

if __name__ == "__main__":
    main()
```

```python
print("05_subtract_7 \n")

def subtract_seven(number):
    return number - 7

def main():
    try:
        num = float(input("Enter a number:  "))
    except ValueError:
        print("Please enter a valid number.")
        return

    result = subtract_seven(num)
    print(f"After subtracting 7, the result is: {result}")

if __name__ == "__main__":
    main()
```
