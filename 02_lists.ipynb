```python
print("01_add_many_number \n")

def add_number(number)->int:
  num:int = 0
  for i in number:
     num += i
  return num

def main():
  numbers:list[int] = [1,2,3,4,5]
  sum :int = add_number(numbers)
  print(sum)

if __name__ == "__main__":
  main()
```

```python
print("02_double_list \n")

def double():
  numbers:list[int] = [2,4,6,8]
  for i in range(len(numbers)):
    numbers[i] *= 2
  print(numbers)

if __name__ == "__main__":
  double()
```
